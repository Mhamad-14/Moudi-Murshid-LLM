## Test plan and expected outcomes

Before executing the code, the expected results are:

| Test input | Expected behavior |
| --- | --- |
| `None` | Catch `ValueError: Enter a non-empty question string.` |
| Blank spaces (`'   '`) | Catch `ValueError: Enter a non-empty question string.` |
| 501-character question | Catch `ValueError: Use at most 500 characters.` |
| Valid question with surrounding spaces | Accept and return the trimmed question |

The first three tests are expected rejections. The code checks the exact error messages and records each rejection without allowing an unhandled exception to stop the run.

```python
# Deliberate, caught errors. These are examples to diagnose, not broken setup.
import json


def validate_public_question(question):
    if not isinstance(question, str) or not question.strip():
        raise ValueError('Enter a non-empty question string.')
    if len(question) > 500:
        raise ValueError('Use at most 500 characters.')
    return question.strip()


def run_error_examples():
    cases = [(None, 'Enter a non-empty question string.'),
             ('   ', 'Enter a non-empty question string.'),
             ('x' * 501, 'Use at most 500 characters.')]
    rows = []
    for value, expected in cases:
        try:
            validate_public_question(value)
        except ValueError as error:
            assert str(error) == expected
            rows.append({'error_type': type(error).__name__, 'message': str(error), 'expected_rejection': True})
        else:
            raise AssertionError('Invalid input was accepted.')
    assert validate_public_question(' How can I check my application status? ') == 'How can I check my application status?'
    result = {'intentional_caught_errors': rows, 'corrected_input': 'accepted', 'checks': 'passed'}
    print(json.dumps(result, indent=2))
    return result



result = run_error_examples()
```

## What the code demonstrates

- `raise` signals invalid input by creating a `ValueError` with a specific message.
- `try` runs the validation on an input that might fail.
- `except ValueError` catches the expected exception so the test can verify its message and continue.
- `assert` checks that the observed behavior matches the expected outcome. The final valid question is accepted and trimmed.

A correctly caught rejection **passes** when invalid input is supposed to be rejected. An unexpected error, a changed error message, or acceptance of an invalid input would be a test failure.

**Limitations:** These examples verify only three intentionally invalid inputs and one valid question. They do not prove that all possible inputs are safe, and they do not modify the core Murshid notebook.
